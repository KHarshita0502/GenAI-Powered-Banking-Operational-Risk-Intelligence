```python
import pandas as pd

path = "/content/drive/MyDrive/Banking Operational Risk Project/Operational_Risk_Events.csv"

df = pd.read_csv(path)

print("Rows and Columns:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("Dataset Shape:", df.shape)

print("\n--- Data Types ---")
print(df.dtypes)

print("\n--- Missing Values ---")
print(df.isnull().sum())

print("\n--- Duplicate Rows ---")
print(df.duplicated().sum())
```

```python
print("--- Event Types ---")
print(df["Event Type"].value_counts())

print("\n--- Process Areas ---")
print(df["Process Area"].value_counts())

print("\n--- Target Values ---")
print(df["Target"].value_counts())

print("\n--- Date Values ---")
print(df["Date"].value_counts().head(10))
```

```python
print("--- Target vs Event Type ---")
print(pd.crosstab(df["Event Type"], df["Target"]))

print("\n--- Target vs Process Area ---")
print(pd.crosstab(df["Process Area"], df["Target"]))

print("\n--- Synth-2022 vs Target ---")
print(pd.crosstab(df["Date"].eq("Synth-2022"), df["Target"]))
```

```python
print(df[
    [
        "Loss Amount (kUSD)",
        "Frequency (Nk)",
        "Severity (Xi)",
        "OpVar (%)",
        "ML",
        "GDP (%)",
        "VIX"
    ]
].describe().round(2))
```

```python
date_check = df[df["Date"] != "Synth-2022"].copy()

date_check["Date"] = pd.to_datetime(date_check["Date"])

print("Actual date records:", len(date_check))
print("Earliest date:", date_check["Date"].min())
print("Latest date:", date_check["Date"].max())
```

```python
df["Date_Clean"] = pd.to_datetime(
    df["Date"],
    errors="coerce"
)

print(df[["Date", "Date_Clean"]].head(10))
```

```python
df["Date_Clean"] = pd.to_datetime(
    df["Date"],
    format="%Y-%m-%d",
    errors="coerce"
)

print(df[["Date", "Date_Clean"]].head(10))
```

```python
print("Valid dates:", df["Date_Clean"].notna().sum())
print("Invalid/missing dates:", df["Date_Clean"].isna().sum())
```

```python
print("Top 10 Highest Loss Events:")

print(
    df[
        [
            "Scenario ID",
            "Event Type",
            "Process Area",
            "Loss Amount (kUSD)",
            "Frequency (Nk)",
            "Severity (Xi)"
        ]
    ]
    .sort_values("Loss Amount (kUSD)", ascending=False)
    .head(10)
)
```

```python
event_loss = (
    df.groupby("Event Type")
      .agg(
          Total_Events=("Scenario ID", "count"),
          Total_Loss_kUSD=("Loss Amount (kUSD)", "sum"),
          Average_Loss_kUSD=("Loss Amount (kUSD)", "mean")
      )
      .sort_values("Total_Loss_kUSD", ascending=False)
)

print(event_loss.round(2))
```

```python
process_loss = (
    df.groupby("Process Area")
      .agg(
          Total_Events=("Scenario ID", "count"),
          Total_Loss_kUSD=("Loss Amount (kUSD)", "sum"),
          Average_Loss_kUSD=("Loss Amount (kUSD)", "mean")
      )
      .sort_values("Total_Loss_kUSD", ascending=False)
)

print(process_loss.round(2))
```

```python
loss_matrix = pd.pivot_table(
    df,
    values="Loss Amount (kUSD)",
    index="Process Area",
    columns="Event Type",
    aggfunc="sum"
)

print(loss_matrix.round(2))
```

```python
frequency_loss = (
    df.groupby("Frequency (Nk)")
      .agg(
          Total_Events=("Scenario ID", "count"),
          Total_Loss_kUSD=("Loss Amount (kUSD)", "sum"),
          Average_Loss_kUSD=("Loss Amount (kUSD)", "mean")
      )
      .reset_index()
)

print(frequency_loss.round(2))
```

```python
correlation = df["Severity (Xi)"].corr(
    df["Loss Amount (kUSD)"]
)

print("Severity vs Loss correlation:", round(correlation, 3))
```

```python
kpis = {
    "Total Risk Events": df["Scenario ID"].count(),
    "Total Financial Loss (kUSD)": df["Loss Amount (kUSD)"].sum(),
    "Average Loss per Event (kUSD)": df["Loss Amount (kUSD)"].mean(),
    "Average Severity": df["Severity (Xi)"].mean(),
    "Average OpVar": df["OpVar (%)"].mean(),
    "Largest Single Loss (kUSD)": df["Loss Amount (kUSD)"].max()
}

for name, value in kpis.items():
    print(f"{name}: {value:.2f}")
```
